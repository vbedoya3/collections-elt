```python
%sql
CREATE OR REPLACE VIEW cobranzas.semantic.vw_cuentas_gestionadas_por_gestor AS
select 
    gc.cuenta_id, 
    ccm.caracterizacion_mensual,
    sc.sub_caracterizacion,
    gc.gestion_diaria,
    rc.gestor
from cobranzas.gold.fact_gestion_cuenta gc
inner join cobranzas.gold.dim_caracterizaciones_mensuales ccm 
on gc.caracterizacion_mensual_id = ccm.caracterizacion_mensual_id
inner join cobranzas.gold.dim_sub_caracterizaciones sc 
on gc.sub_caracterizacion_id = sc.sub_caracterizacion_id
inner join cobranzas.gold.fact_cuenta c
on gc.cuenta_id = c.cuenta_id
inner join cobranzas.gold.dim_responsables_cuenta rc
on c.responsables_cuenta_id = rc.responsables_cuenta_id
```
