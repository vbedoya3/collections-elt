```python
%sql
CREATE OR REPLACE VIEW cobranzas.semantic.vw_clientes_dificil_recuperacion AS
select 
    cl.cliente_id, cl.numero_cliente_unico, bi.bien_inmueble, lc.labora
from cobranzas.gold.dim_clientes cl
inner join cobranzas.gold.dim_bienes_inmuebles bi
on bi.bien_inmueble_id = cl.bien_inmueble_id
left join cobranzas.gold.fact_laboral_clientes lc
on lc.cliente_id = cl.cliente_id
where labora = false
order by lc.labora desc
```
