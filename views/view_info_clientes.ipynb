```python
%sql
CREATE OR REPLACE VIEW cobranzas.semantic.vw_info_clientes AS

select 
    cl.cliente_id, cl.numero_cliente_unico, bi.bien_inmueble, lc.labora, e.empresa, te.tipo_empresa
from cobranzas.gold.dim_clientes cl
inner join cobranzas.gold.dim_bienes_inmuebles bi
on bi.bien_inmueble_id = cl.bien_inmueble_id
left join cobranzas.gold.fact_laboral_clientes lc
on lc.cliente_id = cl.cliente_id
left join cobranzas.gold.dim_empresas e 
on e.empresa_id = lc.empresa_id
inner join cobranzas.gold.dim_tipos_empresas te
on te.tipo_empresa_id = e.tipo_empresa_id
```
