```python
%sql
CREATE OR REPLACE VIEW cobranzas.semantic.vw_distribucion_cartera_judicial AS
with base as (
    select 
        c.saldo_total_honorarios, p.producto
    from cobranzas.gold.fact_cuenta c
    inner join cobranzas.gold.dim_estados_cartera ec
    on c.estado_cartera_id = ec.estado_id
    inner join cobranzas.gold.dim_productos p
    on c.producto_id = p.producto_id
    where ec.estado = 'Judicial'
)


select sum(saldo_total_honorarios),  producto
from base
group by producto
```
