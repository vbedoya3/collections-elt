```python
%sql
CREATE OR REPLACE VIEW cobranzas.semantic.vw_montos_promesas_actual AS

with uniques as (
    select distinct cuenta_id, monto_mensual_l, monto_mensual_usd
    from cobranzas.gold.fact_promesas
where month(fecha_de_pago) = month(current_date))

select sum(monto_mensual_l) monto_mensual_l, sum(monto_mensual_usd) monto_mensual_usd
from uniques

```
