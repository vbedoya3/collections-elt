```python
%sql
DROP CATALOG IF EXISTS cobranzas CASCADE;
-- ----------------------------------------------------------------------------
-- 1. CATALOGO
-- ----------------------------------------------------------------------------
CREATE CATALOG IF NOT EXISTS cobranzas
COMMENT 'Catalogo del proyecto de cobranzas - arquitectura medallion (bronze/silver/gold)';

USE CATALOG cobranzas;


-- ----------------------------------------------------------------------------
-- 2. ESQUEMAS (capas medallion)
-- ----------------------------------------------------------------------------
CREATE SCHEMA IF NOT EXISTS landing
COMMENT 'Capa landing: aloja los volumenes donde llegan los archivos de origen';

CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Capa bronze: datos crudos sin transformar, tal como llegan de origen';

CREATE SCHEMA IF NOT EXISTS silver
COMMENT 'Capa silver: datos limpios, validados y conformados';

CREATE SCHEMA IF NOT EXISTS gold
COMMENT 'Capa gold: datos agregados y listos para consumo de negocio';

CREATE SCHEMA IF NOT EXISTS semantic
COMMENT 'Capa semantic: Vistas para consumo de negocio';

-- ----------------------------------------------------------------------------
-- 3. VOLUMENES (zona de aterrizaje de los CSV de origen)
-- ----------------------------------------------------------------------------
CREATE VOLUME IF NOT EXISTS cobranzas.landing.cuentas
COMMENT 'Zona de aterrizaje de archivos CSV de cuentas';

CREATE VOLUME IF NOT EXISTS cobranzas.landing.promesas
COMMENT 'Zona de aterrizaje de archivos CSV de promesas de pago';
```
