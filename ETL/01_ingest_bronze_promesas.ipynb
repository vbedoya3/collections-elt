```python
%sql
-- ----------------------------------------------------------------------------
-- 5. CARGA BRONZE PROMESAS
-- ----------------------------------------------------------------------------

COPY INTO cobranzas.bronze.promesas
FROM (
  SELECT
    *,
    _metadata.file_name              AS source_file,
    _metadata.file_modification_time AS source_file_modified_at,
    current_timestamp()              AS ingestion_timestamp
  FROM '/Volumes/cobranzas/landing/promesas/'
)
FILEFORMAT = CSV
FORMAT_OPTIONS (
  'header'   = 'true',
  'sep'      = ',',
  'encoding' = 'UTF-8'
);


SELECT count(*) FROM cobranzas.bronze.promesas;
DESCRIBE TABLE cobranzas.bronze.promesas;
```
