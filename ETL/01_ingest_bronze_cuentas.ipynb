```python
%sql
-- ----------------------------------------------------------------------------
-- 5. CARGA BRONZE CUENTAS
-- ----------------------------------------------------------------------------

COPY INTO cobranzas.bronze.cuentas
FROM (
  SELECT
    *,
    _metadata.file_name              AS source_file,
    _metadata.file_modification_time AS source_file_modified_at,
    current_timestamp()              AS ingestion_timestamp
  FROM '/Volumes/cobranzas/landing/cuentas/'
)
FILEFORMAT = CSV
FORMAT_OPTIONS (
  'header'            = 'true',
  'sep'               = ',',
  'encoding'          = 'UTF-8',
  'rescuedDataColumn' = '_rescued_data'
);

-- ----------------------------------------------------------------------------
-- 6. VERIFICACION RAPIDA 
-- ----------------------------------------------------------------------------
 SELECT count(*) FROM cobranzas.bronze.cuentas;
 DESCRIBE TABLE cobranzas.bronze.cuentas;
```
